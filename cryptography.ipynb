```python
def extended_euclidean(a, b):
    """ 

    Implements the Extended Euclidean Algorithm. 

    Returns gcd(a, b) along with coefficients x and y such that: 

    gcd(a, b) = x * a + y * b 

    """
    if b == 0:
        return (a, 1, 0)
    gcd, x1, y1 = extended_euclidean(b, a % b)
    x = y1
    y = x1 - a // b * y1
    return (gcd, x, y)
```

```python
def extended_gcd(a, b):
    """
    Computes the GCD and coefficients x, y using the Extended Euclidean Algorithm.
    """
    if b == 0:
        return (a, 1, 0)
    gcd, x1, y1 = extended_gcd(b, a % b)
    x = y1
    y = x1 - a // b * y1
    return (gcd, x, y)

def generate_coefficients(a, b, num_pairs=10):
    """
    Generates multiple (x, y) pairs that satisfy gcd(a, b) = x * a + y * b.
    """
    gcd, x, y = extended_gcd(a, b)
    coefficient_pairs = []
    for k in range(num_pairs):
        x_new = x + k * (b // gcd)
        y_new = y - k * (a // gcd)
        coefficient_pairs.append((x_new, y_new))
    return coefficient_pairs
```

```python
def repeated_squaring(base, exp, mod):
    """
    Computes (base^exp) % mod using repeated squaring.
    """
    result = 1
    base = base % mod
    while exp > 0:
        if exp % 2 == 1:
            result = result * base % mod
        base = base * base % mod
        exp //= 2
    return result
```

```python
def apply_permutation(SA):
    """
    Applies the permutation function P to the 32-bit output of the S-box substitution.
    Returns the permuted 32-bit binary string.
    """
    if len(SA) != 32:
        raise ValueError('Input to permutation must be exactly 32 bits long.')
    P = [16, 7, 20, 21, 29, 12, 28, 17, 1, 15, 23, 26, 5, 18, 31, 10, 2, 8, 24, 14, 32, 27, 3, 9, 19, 13, 30, 6, 22, 11, 4, 25]
    PSA = ''.join((SA[i - 1] for i in P))
    return PSA
```

```python
def des_compute_L1_R1(L0, R0, PSA):
    """
    Computes the outputs L1 and R1 for a single DES round.
    L1 = R0
    R1 = L0 XOR PSA
    """
    if len(L0) != 32 or len(R0) != 32:
        raise ValueError('L0 and R0 must be exactly 32 bits long.')
    if len(PSA) != 32:
        raise ValueError('PSA must be exactly 32 bits long.')
    L1 = R0
    R1 = XOR(L0, PSA)
    return (L1, R1)
```
